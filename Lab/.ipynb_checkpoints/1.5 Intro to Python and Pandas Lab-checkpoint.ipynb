Anime - Japanese Animation
Manga - Japanese Comic Books

This dataset is from a website called "My Anime List". This website is open to anyone and people can write reviews for each anime on the site. My anime list happens to be the biggest online community for anime and manga reviews. An example of one you might be familiar with is IMDB or RottenTomatoes

Features:
* Title: The main title of the anime.

* Score: The average user rating on MyAnimeList (out of 10).

* Genres: A list of genres associated with the anime

* Episodes: The total number of episodes.

* Synopsis: A brief summary of the anime's plot.

* Popularity: The anime's popularity rank (lower is more popular).

* Members: The number of MyAnimeList users who have the anime on their list.

* Studios: The animation studio(s) that produced the anime.

* Year: The year the anime was released.

This dataset is interesting to me because I have been a long time Anime fan and it intrigues me to see what other people thought of the anime's that I so much love. Seeing other point of view on Anime that I have watch allows for better insight to the trends as well as reccomending me new Animations to watch.


1. What are the top 50 anime?
2. Do longer series have better scores?
3. what studio created the most animes?
4. Over the years, what has been the trend of anime score reviews?
5. What genre is the most popular?

```python
import pandas as pd

#question 1
anime_df = pd.read_csv('anime_dataset.csv')

anime_df.head(50)

# this is a list of the top 50 anime of all time
```

```python
anime_df.tail()

#this is a END of the list of the Top anime(These Animations are still considered good and are not the worst/bad ones) 
```

```python
#question 2
df_filtered = anime_df[(anime_df['episodes'] > 1) & (anime_df['episodes'] < 200)]
correlation = df_filtered['episodes'].corr(anime_df['score'])
print(f"Simple correlation between episodes (2-200) and score: {correlation:.4f}")

# A better way is to bin the episodes and check average scores
episode_bins = [0, 1, 12, 26, 52, 100, 200, anime_df['episodes'].max()]
bin_labels = ['1 (Movie/Special)', '2-12 (Short)', '13-26 (1-2 Cour)', '27-52 (3-4 Cour)', '53-100 (Long)', '101-200 (Very Long)', '201+ (Epic)']
anime_df['episode_bin'] = pd.cut(anime_df['episodes'], bins=episode_bins, labels=bin_labels, right=True)

# Calculate average score for each bin
# Added observed=True to silence the FutureWarning
score_by_episode_bin = anime_df.groupby('episode_bin', observed=True)['score'].mean().reset_index()
score_by_episode_bin.columns = ['Episode Length', 'Average Score']
print("\nAverage score by episode length:")
print(score_by_episode_bin.to_markdown(index=False))
print("\n" + "="*50 + "\n")

# there is no direct corrolation to say that longer series have better scores
```

```python
#question 3

studios_exploded = anime_df.explode('studios')
top_10_studios = studios_exploded['studios'].value_counts().head(10).reset_index()
top_10_studios.columns = ['Studio', 'Anime Count']
print("Top 10 Most Prolific Studios:")
print(top_10_studios.to_markdown(index=False))
print("\n" + "="*50 + "\n")

#The Animation Studio with the most anime out is A-1 Pictures.
```

```python
#question 4

year_counts = anime_df['year'].value_counts()
valid_years = year_counts[year_counts >= 30].index
df_filtered_years = anime_df[anime_df['year'].isin(valid_years) & (anime_df['year'] >= 1980)]

# Group by year and get the mean score
average_score_by_year = df_filtered_years.groupby('year')['score'].mean()

print("Average score by year (from 1980, for years with 30+ releases):")
# Convert series to dataframe for clean markdown printing
average_score_by_year_df = average_score_by_year.tail(10).reset_index()
average_score_by_year_df.columns = ['Year', 'Average Score']
print(average_score_by_year_df.to_markdown(index=False))
print("\n" + "="*50 + "\n")

#Over the last Decade of anime that has come out. there has been a 7.5 Plateau. However with the year of 2023 there was a large spike in ratings
```

```python
#question 5
genres_exploded = anime_df.explode('genres')
top_10_common_genres = genres_exploded['genres'].value_counts().head(10).reset_index()
top_10_common_genres.columns = ['Genre', 'Title Count']
print("Top 10 Most COMMON Genres (by number of titles):")
print(top_10_common_genres.to_markdown(index=False))

# 2. Most POPULAR (highest average 'members' count)
# This shows which genres attract the largest audiences on average
avg_members_by_genre = genres_exploded.groupby('genres')['members'].mean().sort_values(ascending=False).head(10).reset_index()
avg_members_by_genre.columns = ['Genre', 'Average Members']
print("\nTop 10 Most POPULAR Genres (by average members):")
print(avg_members_by_genre.to_markdown(index=False))

print("\n" + "="*50 + "\n")

# the most common genre of anime are 'Action', 'Adventure', 'Fantasy', 'Shounen'. 
# This demographic alignes with the precieved audience of mostly being men in their teen years.
```
