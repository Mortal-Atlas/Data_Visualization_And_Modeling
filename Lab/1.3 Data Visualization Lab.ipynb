```python
import pandas as pd

df = pd.read_csv('spotify-2023.csv', encoding='ISO-8859-1')
print('BPM Statistics:')
bpm_stats = df['bpm'].describe()
print(bpm_stats)

print()

print('Key Statistics:')
key_stats = df['key'].describe()
print(key_stats)

print()

print('Mode Statistics:')
mode_stats = df['mode'].describe()
print(mode_stats)

print()

print('Dance Statistics:')
dance_stats = df['danceability_%'].describe()
df['danceability_%'].value_counts().plot(kind='bar')
print(dance_stats)

print()

print('Energy Statistics:')
energy_stats = df['energy_%'].describe()
df['energy_%'].value_counts().plot(kind='bar')
print(energy_stats)

print()

print('Speechiness Statistics:')
speech_stats = df['speechiness_%'].describe()
df['speechiness_%'].value_counts().plot(kind='bar')
print(speech_stats)

print()

print('Acousticness Statistics:')
acostics_stats = df['acousticness_%'].describe()
df['acousticness_%'].value_counts().plot(kind='bar')
print(acostics_stats)

print()

```
