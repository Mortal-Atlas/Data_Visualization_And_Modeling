```python
import pandas as pd

df = pd.read_csv('spotify-2023.csv', encoding='ISO-8859-1')
print('BPM Statistics:')
bpm_plot = df['bpm'].plot(kind='hist', title='BPM Distribution')
print(bpm_plot)

print()
```

```python

print('Key Statistics:')
key_stats = df['key'].describe()
key_plot = df['key'].value_counts().plot(kind='bar', title='Key Frequency')
print(key_stats)
print(key_plot)

print()
```

```python

print('Mode Statistics:')
mode_stats = df['mode'].describe()
mode_plot = df['mode'].value_counts().plot(kind='pie', title='Mode Distribution')
print(mode_stats)
print(mode_plot)

print()
```

```python

print('Dance Statistics:')
dance_stats = df['danceability_%'].describe()
dance_plot = df['danceability_%'].plot(kind='hist', title='Danceability % Distribution')
print(dance_stats)
print(dance_plot)

print()
```

```python
print('Energy Statistics:')
energy_stats = df['energy_%'].describe()
df['energy_%'].plot(kind='hist', title='Energy % Distribution')
print(energy_stats)

print()
```

```python
print('Speechiness Statistics:')
speech_stats = df['speechiness_%'].describe()
speech_plot = df['speechiness_%'].plot(kind='hist', title='Speechiness % Distribution')
print(speech_stats)
print(speech_plot)

print()
```

```python
print('Acousticness Statistics:')
acoustics_stats = df['acousticness_%'].describe()
acoustic_plot = df['acousticness_%'].plot(kind='hist', title='Acousticness % Distribution')
print(acoustics_stats)
print(acoustic_plot)

print()
```
